## 03 - Styling Your if Statement

In the previous sections, we learned how to write `if` statements and use them with lists. Now we turn to an important topic: **styling your `if` statements**.

Writing clean, readable `if` statements makes your code easier to understand, debug, and maintain. Following a few simple guidelines will improve the quality of your code significantly.

This section is short but important. The examples below show common styling issues and how to fix them.

## 1. General Guidelines for Styling `if` Statements

### Use Consistent Indentation

Always use 4 spaces for indentation. Do not mix tabs and spaces.

```python
# Good
if condition:
    do_something()

# Bad: Inconsistent indentation
if condition:
  do_something()  # 2 spaces
        do_something_else()  # 8 spaces
```

### Put Spaces Around Operators

Use spaces around comparison and logical operators for readability.

```python
# Good
if age >= 18 and has_id:
    print("Allowed")

# Bad: No spaces
if age>=18 and has_id:
    print("Allowed")
```

### Keep Conditions Simple

If a condition is too complex, break it into multiple lines or use variables.

```python
# Bad: Too long
if age >= 18 and has_id and not is_banned and is_active and status == "verified":
    print("Access granted")

# Good: Broken into multiple lines
if (age >= 18 and has_id and not is_banned
        and is_active and status == "verified"):
    print("Access granted")

# Even better: Use variables
is_eligible = age >= 18 and has_id and not is_banned
is_verified = is_active and status == "verified"
if is_eligible and is_verified:
    print("Access granted")
```

## 2. Common Styling Mistakes and How to Fix Them

### Mistake 1: Unnecessary Parentheses

```python
# Bad: Extra parentheses
if (score >= 80):
    print("Good score")

# Good
if score >= 80:
    print("Good score")
```

### Mistake 2: Forgetting Spaces After `if`

```python
# Bad: No space
if condition:
    print("Hello")

# Good
if condition:
    print("Hello")
```

### Mistake 3: Unnecessary `else` After `return`

If a function returns early, you can often omit the `else` for cleaner code.

```python
# Bad: Redundant else
def check_age(age):
    if age >= 18:
        return True
    else:
        return False

# Good: Cleaner
def check_age(age):
    if age >= 18:
        return True
    return False

# Even better
def check_age(age):
    return age >= 18
```

### Mistake 4: Using `if` Without `==` for Boolean Checks

```python
# Bad: Redundant comparison
if is_active == True:
    print("Active")

# Good: Clean
if is_active:
    print("Active")

# Bad: Checking False
if is_deleted == False:
    print("Not deleted")

# Good: Use `not`
if not is_deleted:
    print("Not deleted")
```

## 3. Examples of Clean `if` Statements

### Example 1: Grade Calculation

```python
# Clean version
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Grade: {grade}")
```

### Example 2: Using Meaningful Variable Names

```python
# Bad: Vague names
age = 25
x = True
y = False

if age >= 18 and x and not y:
    print("Welcome")

# Good: Clear names
user_age = 25
has_license = True
is_banned = False

if user_age >= 18 and has_license and not is_banned:
    print("Welcome")
```

### Example 3: Using `if` with Lists

```python
# Clean and readable
available_items = ["laptop", "mouse", "keyboard"]
cart = ["laptop", "tablet"]

for item in cart:
    if item not in available_items:
        print(f"{item} is out of stock")
    else:
        print(f"{item} added to cart")
```

## 4. Comprehensive Clean Code Example

Here is a complete example showing good styling practices for `if` statements.

```python
# Clean if statement example

def process_order(cart, available_items, discounted_items):
    """Process a shopping cart and return the total."""
    
    if not cart:
        print("Cart is empty")
        return 0
    
    total = 0
    
    for item in cart:
        if item not in available_items:
            print(f"{item}: Not available")
            continue
        
        price = available_items[item]
        
        if item in discounted_items:
            total += price * 0.9
            print(f"{item}: ${price} (discounted)")
        else:
            total += price
            print(f"{item}: ${price}")
    
    if total > 100:
        print("Free shipping applied!")
    elif total > 50:
        print("Shipping: $5")
        total += 5
    else:
        print("Shipping: $10")
        total += 10
    
    return total

# Usage
inventory = {"laptop": 1000, "mouse": 50, "keyboard": 80}
discounts = ["laptop"]
cart = ["laptop", "mouse", "tablet"]

total = process_order(cart, inventory, discounts)
print(f"Total: ${total:.2f}")
```

Output:
```
laptop: $1000 (discounted)
mouse: $50
tablet: Not available
Free shipping applied!
Total: $950.00
```

## Summary

- Use **4 spaces** for indentation consistently.
- Put spaces **around operators** and after commas.
- Keep conditions **simple and readable**. Break long conditions into multiple lines.
- Use **meaningful variable names** instead of vague single letters.
- Avoid **unnecessary parentheses** and redundant comparisons.
- Use `if not` instead of `== False` for Boolean checks.
- Break complex logic into **separate variables** for clarity.
- If a function returns early, **omit the `else`**.

Good style is a habit. The more you practice, the more natural it becomes.
