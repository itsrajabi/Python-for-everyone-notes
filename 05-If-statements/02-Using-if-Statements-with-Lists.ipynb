## 02 - Using If Statements with Lists

In the previous section, we learned about `if` statements and how they control the flow of a program. Now we combine `if` statements with lists to handle more realistic scenarios.

Lists are one of the most common data structures in Python. When you combine them with conditional logic, you can build programs that respond intelligently to different data situations – whether it is checking for special items, handling empty lists, or working with multiple lists at the same time.

## 1. Checking for Special Items

When looping through a list, you often want to treat certain items differently. You can use an `if` statement inside a `for` loop to check for special items.

### Example 1: Greeting Users with a Special Welcome

```python
users = ["alice", "bob", "admin", "charlie", "diana"]

for user in users:
    if user == "admin":
        print(f"Hello {user}, would you like to see the admin dashboard?")
    else:
        print(f"Hello {user}, welcome to the site.")
```

Output:
```
Hello alice, welcome to the site.
Hello bob, welcome to the site.
Hello admin, would you like to see the admin dashboard?
Hello charlie, welcome to the site.
Hello diana, welcome to the site.
```

### Example 2: Applying Special Discount

```python
items = ["laptop", "mouse", "keyboard", "monitor"]

for item in items:
    if item == "laptop":
        print(f"{item}: 10% discount applied")
    else:
        print(f"{item}: standard price")
```

Output:
```
laptop: 10% discount applied
mouse: standard price
keyboard: standard price
monitor: standard price
```

### Example 3: Skipping Banned Users

```python
banned_users = ["bob", "charlie"]
users = ["alice", "bob", "charlie", "diana", "eve"]

for user in users:
    if user in banned_users:
        print(f"{user}: Access denied.")
    else:
        print(f"{user}: Access granted.")
```

Output:
```
alice: Access granted.
bob: Access denied.
charlie: Access denied.
diana: Access granted.
eve: Access granted.
```

## 2. Checking That a List Is Not Empty

Before processing a list, it is good practice to check if it is empty. This prevents errors and makes your program more robust.

### Example 1: Processing Orders

```python
orders = ["laptop", "mouse", "keyboard"]

if orders:
    for order in orders:
        print(f"Processing order: {order}")
else:
    print("No orders to process.")
```

Output:
```
Processing order: laptop
Processing order: mouse
Processing order: keyboard
```

### Example 2: Empty List

```python
orders = []

if orders:
    for order in orders:
        print(f"Processing order: {order}")
else:
    print("No orders to process.")
```

Output:
```
No orders to process.
```

### Example 3: Check if Users Exist

```python
active_users = ["alice", "bob", "charlie"]

if active_users:
    print(f"There are {len(active_users)} active users.")
    for user in active_users:
        print(f"  - {user}")
else:
    print("No active users.")
```

Output:
```
There are 3 active users.
  - alice
  - bob
  - charlie
```

### Example 4: Shopping Cart with Empty Check

```python
cart = []

if cart:
    total = 0
    for item in cart:
        total += item["price"]
    print(f"Your total is: ${total}")
else:
    print("Your cart is empty.")
```

Output:
```
Your cart is empty.
```

## 3. Using Multiple Lists

Often you need to compare or combine two lists. This is common in many real-world applications like matching users with permissions, checking inventory, or finding common items.

### Example 1: Checking Availability

```python
available_items = ["laptop", "mouse", "keyboard", "monitor"]
requested_items = ["laptop", "mouse", "tablet"]

for item in requested_items:
    if item in available_items:
        print(f"{item} is in stock.")
    else:
        print(f"{item} is not available.")
```

Output:
```
laptop is in stock.
mouse is in stock.
tablet is not available.
```

### Example 2: Finding Common Items (Intersection)

```python
favorites = ["apple", "banana", "orange"]
available = ["banana", "grape", "orange", "kiwi"]

common_items = []
for item in favorites:
    if item in available:
        common_items.append(item)

if common_items:
    print(f"Available favorites: {common_items}")
else:
    print("None of your favorites are available.")
```

Output:
```
Available favorites: ['banana', 'orange']
```

### Example 3: Users with Permissions

```python
users = ["alice", "bob", "charlie", "diana"]
admins = ["alice", "charlie"]

for user in users:
    if user in admins:
        print(f"{user}: Admin access")
    else:
        print(f"{user}: User access")
```

Output:
```
alice: Admin access
bob: User access
charlie: Admin access
diana: User access
```

## 4. More Useful Techniques with Lists and If Statements

Here are additional patterns that are common when working with lists and conditional logic.

### Technique 1: Filtering a List

Creating a new list containing only items that meet a condition.

```python
numbers = [10, 25, 3, 48, 12, 5, 99, 7]
large_numbers = []

for num in numbers:
    if num > 20:
        large_numbers.append(num)

print(f"Large numbers (>20): {large_numbers}")
```

Output:
```
Large numbers (>20): [25, 48, 99]
```

### Technique 2: Finding the First Match

Sometimes you want to find the first item that meets a condition and stop.

```python
fruits = ["apple", "banana", "orange", "grape"]

for fruit in fruits:
    if fruit[0] == "b":
        print(f"Found a fruit starting with 'b': {fruit}")
        break
```

Output:
```
Found a fruit starting with 'b': banana
```

### Technique 3: Checking All Items Meet a Condition

You can check if all items in a list meet a condition using a flag.

```python
scores = [85, 92, 78, 90, 88]
all_passing = True

for score in scores:
    if score < 70:
        all_passing = False
        break

if all_passing:
    print("All students passed.")
else:
    print("At least one student failed.")
```

Output:
```
All students passed.
```

### Technique 4: Checking if Any Item Meets a Condition

```python
numbers = [1, 2, 3, 4, 5]
has_even = False

for num in numbers:
    if num % 2 == 0:
        has_even = True
        break

if has_even:
    print("The list contains at least one even number.")
else:
    print("No even numbers found.")
```

Output:
```
The list contains at least one even number.
```

### Technique 5: Removing Items Conditionally

```python
tasks = ["Write code", "Test", "Deploy", "Review", "Fix bugs"]

# Remove all tasks with length less than 5 characters
tasks_to_keep = []
for task in tasks:
    if len(task) >= 5:
        tasks_to_keep.append(task)

print(f"Remaining tasks: {tasks_to_keep}")
```

Output:
```
Remaining tasks: ['Write code', 'Deploy', 'Review', 'Fix bugs']
```

## 5. Comprehensive Example

Let us combine everything into a realistic scenario: managing an online store order system.

### Scenario: Order Processing System

We have:
- A list of available products with their prices
- A shopping cart (items the customer wants to buy)
- A list of discounted items (10% off)
- A list of items that are out of stock

We need to:
1. Check if the cart is empty
2. For each item, check availability
3. Apply discounts to eligible items
4. Calculate and display the total

```python
# Data
products = {
    "laptop": 1000,
    "mouse": 50,
    "keyboard": 80,
    "monitor": 300,
    "tablet": 400
}

cart = ["laptop", "mouse", "tablet", "keyboard"]
discounted_items = ["laptop", "monitor"]
out_of_stock = ["tablet"]

# Check if cart is empty
if not cart:
    print("Your cart is empty. Please add items.")
else:
    total = 0
    print("=== Order Summary ===\n")
    
    for item in cart:
        # Check if item is in stock
        if item in out_of_stock:
            print(f"❌ {item}: Out of stock - removed from order")
            continue
        
        # Check if item exists in products
        if item not in products:
            print(f"⚠️ {item}: Unknown product")
            continue
        
        # Calculate price with discount
        price = products[item]
        if item in discounted_items:
            discounted_price = price * 0.9
            print(f"✅ {item}: ${price} -> ${discounted_price:.2f} (10% off)")
            total += discounted_price
        else:
            print(f"✅ {item}: ${price}")
            total += price
    
    # Display total
    print(f"\nTotal amount: ${total:.2f}")
    
    # Check for free shipping
    if total > 500:
        print("🎉 Free shipping applied!")
    else:
        print(f"Add ${500 - total:.2f} more for free shipping.")
```

Output:
```
=== Order Summary ===

✅ laptop: $1000 -> $900.00 (10% off)
✅ mouse: $50
❌ tablet: Out of stock - removed from order
✅ keyboard: $80

Total amount: $1030.00
🎉 Free shipping applied!
```

## Summary

- Use `if` statements inside loops to check for **special items**.
- Always check if a list is **empty** before processing it (`if list:`).
- Use `in` and `not in` to compare **multiple lists** (e.g., availability, permissions).
- **Filtering** creates a new list with items that meet a condition.
- Use a **flag variable** to check if all or any items meet a condition.
- **Remove items** from a list conditionally by building a new list.
- Combine all these techniques to handle realistic scenarios like shopping carts, user management, and inventory.
