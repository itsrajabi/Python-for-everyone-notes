# Chapter 06: Dictionaries
## 00 - Working with Dictionaries

In the previous chapters, we worked with lists – collections of items stored in a specific order. Now we turn to **dictionaries**, one of Python's most powerful and versatile data structures.

A **dictionary** stores **key-value pairs**. Each key is connected to a value, and you can use the key to access the value. This is similar to a real-world dictionary where you look up a word (key) to find its definition (value).

Dictionaries are used to represent real-world objects with properties, such as a person with a name, age, and email address. They are also used for mapping, configuration settings, counting, and many other tasks.

## 1. A Simple Dictionary

A dictionary is defined using curly braces `{}` with key-value pairs separated by colons `:`. The keys and values are separated by commas.

### Example 1: A Simple Person Dictionary

```python
person = {
    "first_name": "Alice",
    "last_name": "Smith",
    "age": 30,
    "city": "New York"
}

print(person)
```

Output:
```
{'first_name': 'Alice', 'last_name': 'Smith', 'age': 30, 'city': 'New York'}
```

### Example 2: A Simple Dictionary on One Line

```python
favorite_color = {"Alice": "blue", "Bob": "green", "Charlie": "red"}
print(favorite_color)
```

Output:
```
{'Alice': 'blue', 'Bob': 'green', 'Charlie': 'red'}
```

### Example 3: Dictionary with Mixed Types

```python
car = {
    "brand": "Toyota",
    "model": "Camry",
    "year": 2020,
    "is_electric": False,
    "colors": ["red", "blue", "white"]
}

print(car)
```

Output:
```
{'brand': 'Toyota', 'model': 'Camry', 'year': 2020, 'is_electric': False, 'colors': ['red', 'blue', 'white']}
```

## 2. Accessing Values in a Dictionary

To access a value, use the key in square brackets `[]`.

### Example 1: Accessing Values

```python
person = {
    "first_name": "Alice",
    "last_name": "Smith",
    "age": 30,
    "city": "New York"
}

print(person["first_name"])
print(person["age"])
print(person["city"])
```

Output:
```
Alice
30
New York
```

### Example 2: Using Variables as Keys

```python
user = {
    "username": "alice123",
    "email": "alice@example.com"
}

key = "username"
print(user[key])  # alice123
```

Output:
```
alice123
```

### Example 3: Accessing Values in a Nested Dictionary

```python
user = {
    "name": "Alice",
    "address": {
        "street": "123 Main St",
        "city": "Boston",
        "zip": "02101"
    }
}

print(user["address"]["city"])  # Boston
```

Output:
```
Boston
```

## 3. Starting with an Empty Dictionary

You can start with an empty dictionary and add key-value pairs later. This is useful when you are building data dynamically.

### Example 1: Building a Dictionary from Scratch

```python
person = {}
person["name"] = "Alice"
person["age"] = 30
person["city"] = "New York"

print(person)
```

Output:
```
{'name': 'Alice', 'age': 30, 'city': 'New York'}
```

### Example 2: Building a Shopping Cart

```python
cart = {}
cart["item1"] = "laptop"
cart["item2"] = "mouse"
cart["item3"] = "keyboard"

print(cart)
```

Output:
```
{'item1': 'laptop', 'item2': 'mouse', 'item3': 'keyboard'}
```

### Example 3: Building with User Input (Conceptual)

```python
user = {}
user["name"] = input("Enter your name: ")  # Would require input
user["age"] = int(input("Enter your age: "))

print(user)
```

## 4. Removing Key-Value Pairs

You can remove a key-value pair using the `del` statement or the `pop()` method.

### Example 1: Using `del`

```python
person = {
    "name": "Alice",
    "age": 30,
    "city": "New York",
    "email": "alice@example.com"
}

del person["email"]
print(person)
```

Output:
```
{'name': 'Alice', 'age': 30, 'city': 'New York'}
```

### Example 2: Using `pop()`

The `pop()` method removes a key-value pair and returns the removed value.

```python
person = {
    "name": "Alice",
    "age": 30,
    "city": "New York"
}

removed_age = person.pop("age")
print(f"Removed: {removed_age}")
print(person)
```

Output:
```
Removed: 30
{'name': 'Alice', 'city': 'New York'}
```

### Example 3: Removing a Non-Existent Key (With Default)

```python
person = {"name": "Alice", "age": 30}
value = person.pop("city", "Not found")
print(value)  # Not found
print(person)  # {'name': 'Alice', 'age': 30}
```

Output:
```
Not found
{'name': 'Alice', 'age': 30}
```

### Example 4: Using `clear()` to Remove All Items

```python
person = {"name": "Alice", "age": 30}
person.clear()
print(person)  # {}
```

Output:
```
{}
```

## 5. A Dictionary of Similar Objects

A dictionary can contain multiple objects with the same structure. This is useful for storing data about multiple people, products, or items.

### Example 1: List of People

```python
people = {
    "alice": {
        "first_name": "Alice",
        "last_name": "Smith",
        "age": 30
    },
    "bob": {
        "first_name": "Bob",
        "last_name": "Johnson",
        "age": 25
    },
    "charlie": {
        "first_name": "Charlie",
        "last_name": "Brown",
        "age": 35
    }
}

print(people["alice"]["first_name"])
print(people["bob"]["age"])
```

Output:
```
Alice
25
```

### Example 2: List of Products

```python
products = {
    "p001": {"name": "Laptop", "price": 1000, "in_stock": True},
    "p002": {"name": "Mouse", "price": 50, "in_stock": True},
    "p003": {"name": "Keyboard", "price": 80, "in_stock": False}
}

for product_id, details in products.items():
    status = "in stock" if details["in_stock"] else "out of stock"
    print(f"{details['name']}: ${details['price']} ({status})")
```

Output:
```
Laptop: $1000 (in stock)
Mouse: $50 (in stock)
Keyboard: $80 (out of stock)
```

### Example 3: Nested Loop Through Similar Objects

```python
users = {
    "user1": {"name": "Alice", "email": "alice@example.com"},
    "user2": {"name": "Bob", "email": "bob@example.com"}
}

for username, info in users.items():
    print(f"Username: {username}")
    print(f"  Name: {info['name']}")
    print(f"  Email: {info['email']}")
    print()
```

Output:
```
Username: user1
  Name: Alice
  Email: alice@example.com

Username: user2
  Name: Bob
  Email: bob@example.com
```

## 6. Using `get()` to Access Values

When you use square brackets to access a value, Python raises a `KeyError` if the key does not exist. The `get()` method handles this gracefully by returning `None` or a default value.

### Example 1: `get()` with Existing Key

```python
person = {"name": "Alice", "age": 30}
print(person.get("name"))  # Alice
```

Output:
```
Alice
```

### Example 2: `get()` with Non-Existent Key

```python
person = {"name": "Alice", "age": 30}
print(person.get("city"))  # None
print(person.get("city", "Unknown"))  # Unknown
```

Output:
```
None
Unknown
```

### Example 3: Using `get()` in Practice

```python
user_input = {"username": "alice123"}
email = user_input.get("email", "Not provided")
print(f"Email: {email}")
```

Output:
```
Email: Not provided
```

### Example 4: Comparing `[]` and `get()`

```python
person = {"name": "Alice", "age": 30}

# Using [] - raises error if key doesn't exist
print(person["name"])  # Works

# Using get() - safe
print(person.get("city", "Not specified"))  # Not specified
```

Output:
```
Alice
Not specified
```

## 7. More Examples

Let us look at several more practical examples combining everything we have learned.

### Example 1: Inventory Management

```python
inventory = {}

# Add items to inventory
inventory["laptop"] = 10
inventory["mouse"] = 50
inventory["keyboard"] = 30
inventory["monitor"] = 15

print("Current inventory:")
for item, quantity in inventory.items():
    print(f"  {item}: {quantity}")

# Update quantity
inventory["mouse"] = 45

# Remove an item
removed = inventory.pop("monitor", 0)
print(f"\nRemoved monitor (quantity: {removed})")

print("\nUpdated inventory:")
for item, quantity in inventory.items():
    print(f"  {item}: {quantity}")
```

Output:
```
Current inventory:
  laptop: 10
  mouse: 50
  keyboard: 30
  monitor: 15

Removed monitor (quantity: 15)

Updated inventory:
  laptop: 10
  mouse: 45
  keyboard: 30
```

### Example 2: User Profiles with Contact Info

```python
user_profiles = {
    "alice": {
        "full_name": "Alice Smith",
        "email": "alice@example.com",
        "phone": "123-456-7890"
    },
    "bob": {
        "full_name": "Bob Johnson",
        "email": "bob@example.com",
        "phone": "987-654-3210"
    }
}

def get_user_info(username, field):
    """Get a specific field from a user profile."""
    user = user_profiles.get(username)
    if user:
        return user.get(field, "Not provided")
    return "User not found"

print(get_user_info("alice", "email"))
print(get_user_info("bob", "phone"))
print(get_user_info("charlie", "email"))
```

Output:
```
alice@example.com
987-654-3210
User not found
```

### Example 3: Building a Dictionary from Two Lists

```python
students = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]

student_scores = {}
for i in range(len(students)):
    student_scores[students[i]] = scores[i]

print(student_scores)

# Access using get()
print(f"Alice's score: {student_scores.get('Alice', 0)}")
print(f"David's score: {student_scores.get('David', 'Not found')}")
```

Output:
```
{'Alice': 85, 'Bob': 92, 'Charlie': 78}
Alice's score: 85
David's score: Not found
```

## Summary

- A **dictionary** stores key-value pairs using curly braces `{}`.
- Access values using `dictionary[key]` (raises `KeyError` if key doesn't exist).
- Use `get(key, default)` to safely access values.
- Start with an empty dictionary `{}` and add key-value pairs dynamically.
- Remove key-value pairs with `del` (removes without returning) or `pop()` (removes and returns).
- Use `clear()` to remove all items from a dictionary.
- Store **similar objects** by creating nested dictionaries.
- Dictionaries are powerful for mapping data, configurations, and real-world objects.
